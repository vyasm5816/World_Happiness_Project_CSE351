```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load datasets
data_2015 = pd.read_csv("world_happiness/2015.csv")
data_2016 = pd.read_csv("world_happiness/2016.csv")
data_2017 = pd.read_csv("world_happiness/2017.csv")
data_2018 = pd.read_csv("world_happiness/2018.csv")
data_2019 = pd.read_csv("world_happiness/2019.csv")

# Store datasets
datasets = {
    2015: data_2015,
    2016: data_2016,
    2017: data_2017,
    2018: data_2018,
    2019: data_2019
}

# Inspect datasets
for year, df in datasets.items():
    print(f"\n===== {year} =====")
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())
    display(df.head())
```
